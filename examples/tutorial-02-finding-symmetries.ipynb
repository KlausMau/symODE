# Finding symmetries

We use the van der Pol system to set up the polynomial equations for a linear symmetry matrix. The symmetry condition is

$$M f(x) - f(M x) = 0.$$

This tutorial only constructs the symbolic equations and extracts their monomial coefficients.

```python
from symode.dynamical_system import DynamicalSystem
from symode.util import get_polynomial_coefficients
```

## The van der Pol system

```python
system = DynamicalSystem("van_der_pol")
```

## Construct the symmetry equations

The public `DynamicalSystem` method constructs the expanded vector $M f(x) - f(M x)$ and returns the unknown matrix entries separately.

```python
symmetry_equations, unknowns = system.get_symmetry_equations()
symmetry_equations
```

## Extract polynomial coefficients

Each component equation must vanish for every $x$ and $y$. The utility function turns this into equations for the matrix entries by collecting the coefficients of all monomials.

```python
x, y = system.get_variables()
all_coefficient_equations = [
    coefficient
    for equation in symmetry_equations.values()
    for coefficient in get_polynomial_coefficients(equation, [x, y]).values()
]
all_coefficient_equations
```

```python
from sympy import groebner, solve

groebner_basis = groebner(all_coefficient_equations, order="grevlex")
groebner_basis
```

```python
solve(groebner_basis)
```
