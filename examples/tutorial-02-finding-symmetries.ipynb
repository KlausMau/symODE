# Finding symmetries

We use the van der Pol system to set up the polynomial equations for a linear symmetry matrix. The symmetry condition is

$$M f(x) - f(M x) = 0.$$

This tutorial constructs the symbolic equations and extracts their monomial coefficients.

```python
from symode.dynamical_system import DynamicalSystem
from symode.util import get_polynomial_coefficients
```

## The van der Pol system

```python
system = DynamicalSystem("van_der_pol")
```

## Construct the symmetry equations

The public `DynamicalSystem` method constructs the expanded vector $M f(x) - f(M x)$ for an injected matrix ansatz.

```python
from sympy import Matrix, symbols

n = len(system.get_variables())
ansatz = Matrix(n, n, lambda i, j: symbols(f"m_{i}_{j}"))
commutator = system.get_commutator_to_linear_transformation(ansatz)
commutator
```

## Extract polynomial coefficients

Each component equation must vanish for every $x$ and $y$. The utility function turns this into equations for the matrix entries by collecting the coefficients of all monomials.

```python
all_commutator_monomial_coefficients = [
    coefficient
    for commutator_component in commutator.values()
    for coefficient in get_polynomial_coefficients(
        commutator_component, list(system.get_variables())
    ).values()
]
all_commutator_monomial_coefficients
```

```python
from sympy import groebner, solve

groebner_basis = groebner(all_commutator_monomial_coefficients, ansatz, order="grevlex")
groebner_basis
```

```python
all_solutions = solve(groebner_basis)
all_solutions
```

```python
ansatz.subs(all_solutions[0])
```
